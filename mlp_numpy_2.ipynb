```python
import numpy as np
```

```python
#Funções de ativação e derivadas

def sigmoid(z):
    """Função de ativação para as camadas escondidas.
    A função sigmoide é definida por f(x) = 1 / 1+e^-x"""
    return 1.0 / (1.0 + np.exp(-z))

def sigmoid_deriv(z):
    """Derivada da função sigmoid para o Backpropagation. É usada na backpropagation para retro propagar o erro a partir da saída / Loss passando de volta por cada nó, neurônio ou peso da rede neural.
    Pode ser demonstrada aplicando a regra da cadeia da função sigmóide =
    função sigmoide x derivada da função sigmoide. Após os passos de derivação chegamos na função final σ(x)(1 - σ(x))"""
    s = sigmoid(z)
    return s * (1.0 - s)

def linear(z):
    """Função de ativação linear para a saída (Regressão) - retorna sempre a própria constante. O objetivo é prever os números reais contínuos, e não achatá-los entre 1 e 0."""
    return z

def linear_deriv(z):
    """Derivada da função linear - como é uma constante, a derivada é sempre 1."""
    return np.ones_like(z)

print("Funções de ativação carregadas com sucesso!")
```

```python
# CONFIGURAÇÃO INICIAL DA REDE (Igual ao Slide 4)
# Arquitetura: 2 (Entrada) -> 3 (Oculta 1) -> 3 (Oculta 2) -> 2 (Saída) -> Loss

# Entrada (x) baseados nos slides
x = np.array([[0.3],
              [0.5]])

# Valores esperados para o alvo (y), conforme o slide 10: "Vamos considerar que os valores esperados de saída para os dois neurônios da saída sejam: y1 = 1 e y2 = 0". No final, esse valor será comparado com a saída real (a3)
y = np.array([[1.0],
              [0.0]])

# Pesos e Biases da Camada 1 (Escondida)
W1 = np.array([[0.1, 0.2],
               [0.1, 0.8],
               [0.5, 0.9]])
b1 = np.array([[0.2], [0.4], [0.6]])

# Pesos e Biases da Camada 2 (Escondida)
W2 = np.array([[0.3, 0.4, 0.4],
               [0.7, 0.4, 0.4],
               [0.6, 0.5, 0.3]])
b2 = np.array([[0.4], [0.3], [0.7]])

# Pesos e Biases da Camada 3 (Saída)
W3 = np.array([[0.2, 0.2, 0.6],
               [0.7, 0.3, 0.8]])
b3 = np.array([[0.5], [0.6]])

# Tamanho do passo para corrigir os erros / ajustar os pesos
taxa_aprendizagem = 0.1

print(" Rede configurada! \n")
```

```python
#Forward pass - passagem para frente

# Aqui ele vai calcular a soma ponderada dos neurônios: z = (w1.a1) + (w2.a2) + (w3.a3) + b. Porém isso deverá ser feito na rede toda

# Neste caso a computação se desdobra o np.dot(w1, x) + b1 por debaixo dos panos par calcular o z dos 3 neurônio da camada 2 simultaneamente, usando os valores do código extraídos do slide, dessa forma:
# Neurônio 1: z1 = (w1,1.x1) + (w1,2.x2) + b1
#             z1 = (0,1 . 0,3) + (0,2 . 0,5) + 0,2

# Neurônio 2: z2 = (w2,1.x1) + (w2,2.x2) + b2
#             z2 = (0,1 . 0,3) + (0,8 . 0,5) + 0,4

# Neurônio 3: z3 = (w3,1.x1) + (w3,2.x2) + b3
#             z3 = (0,5 . 0,3) + (0,9 . 0,5) + 0,6

# Então, ao invés de fazer o Python calcular linha por linha, o método 'np.dot' do numpy agrupa tudo e devolve os três resultados de uma só vez.

print(" EXECUÇÃO PASSO A PASSO (1ª ÉPOCA) \n")

print("--- FORWARD PASS (Propagação para Frente) ---")
print("Equação geral da camada: z = W*x + b  ->  a = funcao_ativacao(z)\n")

# Camada 1
print("=> CAMADA 1 (Escondida)")
z1 = np.dot(W1, x) + b1
print(f"Soma Ponderada (z1 = W1*x + b1):\n{z1}\n")
a1 = sigmoid(z1)
print(f"Ativação (a1 = sigmoid(z1)):\n{a1}\n")

# Camada 2
print("=> CAMADA 2 (Escondida)")
z2 = np.dot(W2, a1) + b2
print(f"Soma Ponderada (z2 = W2*a1 + b2):\n{z2}\n")
a2 = sigmoid(z2)
print(f"Ativação (a2 = sigmoid(z2)):\n{a2}\n")

# Camada 3 (Saída Linear para Regressão)
print("=> CAMADA 3 (Saída)")
z3 = np.dot(W3, a2) + b3
print(f"Soma Ponderada (z3 = W3*a2 + b3):\n{z3}\n")
a3 = linear(z3)
print(f"Saída Final (a3 = linear(z3)):\n{a3}\n")

# Cálculo do Erro (Loss - MLE / Erro Quadrático)
loss = 0.5 * np.sum((a3 - y)**2)
print("=> CÁLCULO DO ERRO (LOSS)")
print(f"Alvo esperado (y):\n{y}")
print(f"Fórmula: 0.5 * soma((a3 - y)^2)")
print(f"Loss (Erro Máxima Verossimilhança): {loss:.6f}\n")
```

```python
# Backpropagation

print("--- BACKPROPAGATION (Cálculo dos Gradientes via Regra da Cadeia) ---\n")

# PASSO 1: Derivada da Loss em relação à saída final. É onde o fluxo de correção começa.
dL_da3 = a3 - y
print("=> DERIVADA DO ERRO")
print("Derivada da Loss em relação à saída (dL/da3 = a3 - y):\n", dL_da3, "\n")

# --- CAMADA 3 (Saída) ---
print("=> GRADIENTES DA CAMADA 3")
# Passo 2: dL/dz3 = dL/da3 * derivada_ativacao(z3) - multiplica o erro da saída pela derivada da função de ativação da camada 3. Como na saída usamos a função linear, a derivada dela é sempre 1 => o erro passa por ela sem sofrer nenhuma alteração
dL_dz3 = dL_da3 * linear_deriv(z3)
print("Erro propagado para a soma ponderada (dL/dz3 = dL/da3 * derivada_linear):\n", dL_dz3, "\n")

# Passo 3 e 4 - gradiente de bias e pesos.
# Passo 3: dL/db3 = dL/dz3 => o gradiente do bias é o próprio erro dL/dz3, pois o bias entra somando direto sem multiplicador
dL_db3 = dL_dz3

# Passo 4: dL/dW3 = dL/dz3 * a2_matriz_transposta => o gradiente de pesos w3 é a multiplicação do erro da saída pelo sinal que veio da camada 2 (a2.T)
dL_dW3 = np.dot(dL_dz3, a2.T)
print("Gradiente dos pesos da Camada 3 (dL/dW3 = dL/dz3 * a2.T):\n", dL_dW3, "\n")

# Passo 5: Propagar erro para a camada anterior (dL/da2) multiplicando a matriz transposta dos pesos da camada 3 (w3.T) pelo erro dl/dz3. É como se a camada 3 mostrasse para a camada 2 que já calculou a parcela de erro, passando essa parcela para a camada 2 ajustar os pesos também.
dL_da2 = np.dot(W3.T, dL_dz3)
print("Erro propagado para a Camada 2 (dL/da2 = W3.T * dL/dz3):\n", dL_da2, "\n")

# --- CAMADA 2 (Escondida) ---
print("=> GRADIENTES DA CAMADA 2")
# recebando o erro da camada anterior para fazer o mesmo
dL_dz2 = dL_da2 * sigmoid_deriv(z2)
print("Erro na soma ponderada da Camada 2 (dL/dz2 = dL/da2 * derivada_sigmoid):\n", dL_dz2, "\n")
# ajuste do peso e bias da camada 2 para repassar para a 1
dL_db2 = dL_dz2
dL_dW2 = np.dot(dL_dz2, a1.T)
print("Gradiente dos pesos da Camada 2 (dL/dW2):\n", dL_dW2, "\n")
dL_da1 = np.dot(W2.T, dL_dz2)

# --- CAMADA 1 (Escondida) ---
print("=> GRADIENTES DA CAMADA 1")
dL_dz1 = dL_da1 * sigmoid_deriv(z1)
dL_db1 = dL_dz1
dL_dW1 = np.dot(dL_dz1, x.T)
print("Gradiente dos pesos da Camada 1 (dL/dW1):\n", dL_dW1, "\n")
# Não é necessário calcular dL_da0 (entrada), pois não ajustamos os dados de entrada.
```

```python
#Atualização dos pesos (gradiente descendente)

# Depois de passar por todos os 5 passos em todas as camadas e guardar os gradientes (dL_dW e dL_db), a rede atualiza a sua memória.

print("--- ATUALIZAÇÃO DOS PESOS (Gradiente Descendente) ---")
print("Fórmula base: Novo_Peso = Peso_Atual - (taxa_aprendizagem * Gradiente)\n")

W3 = W3 - taxa_aprendizagem * dL_dW3
b3 = b3 - taxa_aprendizagem * dL_db3

W2 = W2 - taxa_aprendizagem * dL_dW2
b2 = b2 - taxa_aprendizagem * dL_db2

W1 = W1 - taxa_aprendizagem * dL_dW1
b1 = b1 - taxa_aprendizagem * dL_db1

print("Pesos atualizados com sucesso!\n")
print("Exemplo de exibição - Novos Pesos da Camada 3 (W3):\n", W3, "\n")
print("Exemplo de exibição - Novos Biases da Camada 3 (b3):\n", b3, "\n")
```

```python
#Ciclo de treino rápido para provar a aprendizagem

print(" INICIANDO TREINO DE 1000 ÉPOCAS (Omissão de passos) \n")

epocas = 1000
for epoca in range(epocas):
    # Forward Pass
    z1 = np.dot(W1, x) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(W2, a1) + b2
    a2 = sigmoid(z2)
    z3 = np.dot(W3, a2) + b3
    a3 = linear(z3)

    # Loss
    loss = 0.5 * np.sum((a3 - y)**2)

    # Backpropagation
    dL_da3 = a3 - y

    dL_dz3 = dL_da3 * linear_deriv(z3)
    dL_dW3 = np.dot(dL_dz3, a2.T)
    dL_db3 = dL_dz3
    dL_da2 = np.dot(W3.T, dL_dz3)

    dL_dz2 = dL_da2 * sigmoid_deriv(z2)
    dL_dW2 = np.dot(dL_dz2, a1.T)
    dL_db2 = dL_dz2
    dL_da1 = np.dot(W2.T, dL_dz2)

    dL_dz1 = dL_da1 * sigmoid_deriv(z1)
    dL_dW1 = np.dot(dL_dz1, x.T)
    dL_db1 = dL_dz1

    # Update
    W3 -= taxa_aprendizagem * dL_dW3
    b3 -= taxa_aprendizagem * dL_db3
    W2 -= taxa_aprendizagem * dL_dW2
    b2 -= taxa_aprendizagem * dL_db2
    W1 -= taxa_aprendizagem * dL_dW1
    b1 -= taxa_aprendizagem * dL_db1

    if (epoca+1) % 200 == 0:
        print(f"Época {epoca+1:4d} | Loss: {loss:.6f}")

print("\n--- RESULTADO FINAL ---")
print("Saída Prevista (deve estar próxima de [1.0, 0.0]):")
print(np.round(a3, 6))
```
